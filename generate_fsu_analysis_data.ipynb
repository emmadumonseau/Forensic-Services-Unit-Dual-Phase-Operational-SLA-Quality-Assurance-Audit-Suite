```python
import os
import numpy as np
import pandas as pd

def generate_casework_data():
    # Define target desktop directory
    output_dir = r"C:\Users\Edumo\Desktop\Forensic Science Unit Analysis"
    os.makedirs(output_dir, exist_ok=True)

    # Set seed so dataset generation is reproducible
    np.random.seed(2026)
    total_records = 1000

    # SPPD Forensic Services Unit operational disciplines
    units = [
        'DNA', 'Drug Chemistry', 'Crime Scene', 
        'Latent Prints', 'Firearms', 'Toxicology', 'Trace Evidence'
    ]
    unit_weights = [0.25, 0.25, 0.15, 0.15, 0.10, 0.05, 0.05]

    priorities = ['Routine', 'Priority (Urgent)', 'Court Subpoena / Trial']
    priority_weights = [0.65, 0.25, 0.10]

    # Build primary evidence log attributes
    raw_log = {
        'Case_Number': [f"STP-26-{np.random.randint(10000, 99999)}" for _ in range(total_records)],
        'Item_ID': [f"EVD-{i:04d}" for i in range(1, total_records + 1)],
        'Assigned_Unit': np.random.choice(units, total_records, p=unit_weights),
        'Priority_Level': np.random.choice(priorities, total_records, p=priority_weights),
        'Date_Received': pd.date_range(start='2026-01-01', periods=total_records, freq='7h'),
        'Chain_of_Custody_Verified': np.random.choice([1, 0], total_records, p=[0.98, 0.02]),
        'QC_Calibration_Passed': np.random.choice([1, 0], total_records, p=[0.96, 0.04]),
        'Peer_Review_Status': np.random.choice(
            ['Verified & Signed', 'Pending Review', 'Active Analysis', 'QC Hold'], 
            total_records, 
            p=[0.70, 0.15, 0.10, 0.05]
        )
    }

    df = pd.DataFrame(raw_log)

    # Standard SLA turnaround targets by unit (in days)
    sla_targets = {
        'Crime Scene': 7,
        'Drug Chemistry': 14,
        'Latent Prints': 18,
        'Toxicology': 20,
        'Firearms': 21,
        'DNA': 25,
        'Trace Evidence': 30
    }
    df['Target_SLA_Days'] = df['Assigned_Unit'].map(sla_targets)

    # Model realistic processing timelines around targets with normal variance
    processing_days = df['Target_SLA_Days'] + np.random.normal(loc=1.5, scale=4.0, size=total_records).round(1)
    df['Actual_Turnaround_Days'] = np.maximum(processing_days, 1.0)

    # Export clean dataset to target folder
    output_filepath = os.path.join(output_dir, 'simulated_fsu_casework_audit.csv')
    df.to_csv(output_filepath, index=False)
    print(f"Generated {total_records:,} casework records with turnaround metrics -> saved to '{output_filepath}'")

if __name__ == "__main__":
    generate_casework_data()
```
