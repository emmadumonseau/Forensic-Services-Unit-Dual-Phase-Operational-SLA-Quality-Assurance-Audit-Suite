```python
import pandas as pd
import numpy as np
import os

# 1. Environment Initialization
output_dir = r"C:\Users\Edumo\Desktop\Forensic Science Unit Analysis"
os.makedirs(output_dir, exist_ok=True)

np.random.seed(2026)
n_cases = 1000

forensic_units = ['Crime Scene', 'DNA', 'Drug Chemistry', 'Latent Prints', 'Firearms', 'Toxicology', 'Trace Evidence']

print(f"Generating Technical Casework Dataset...")

# 2. Build Technical Casework Dictionary
fsu_technical_data = {
    # Base Identifiers 
    'Case_Number': [f"STP-26-{np.random.randint(10000, 99999)}" for _ in range(n_cases)],
    'Item_ID': [f"TEC-{i:04d}" for i in range(1, n_cases +1)],
    'Assigned_Unit': np.random.choice(forensic_units, size=n_cases),

    # Crime Scene & Evidence Logistics 
    'Packaging_Integrity_Status': np.random.choice(
        ['Sealed/Intact', 'Packaging Discrepancy'],
        n_cases,
        p=[0.97, 0.03]
    ),
    'Transport_Temp_Controlled': np.random.choice(
        ['1', '0'],
        n_cases,
        p=[0.95, 0.05]
    ),
    'Chain_of_Custody_Intact': np.random.choice(
        ['1', '0'],
        n_cases,
        p=[0.98, 0.02]
    ),

    # Biological / DNA Screening & CODIS
    'DNA_Quant_Yield_ng_ul': np.round(np.random.exponential(scale=15.5, size=n_cases), 2),
    'CODIS_Eligible': np.random.choice([1, 0], size=n_cases, p=[0.40, 0.60]),
    'CODIS_Upload_Days': np.random.randint(2, 25, size=n_cases),

    # Controlled Substances & Drug Chemistry
    'Confirmatory_Instrument': np.random.choice(
        ['GC-MS', 'FTIR', 'LC-MS/MS', 'N/A (Non-Drug)'],
        n_cases,
        p=[0.30, 0.20, 0.10, 0.40]
    ),
    'QC_Blank_Check_Passed': np.random.choice([1, 0], size=n_cases, p=[0.98, 0.02]),
    'Measurement_Uncertainty_Pct': np.round(np.random.uniform(0.5, 3.2, size=n_cases), 2),

    # Firearms, Ballistics & NIBIN Operations
    'NIBIN_Eligible': np.random.choice([1, 0], size=n_cases, p=[0.25, 0.75]),
    'NIBIN_Correlation_Days': np.random.randint(1, 7, size=n_cases),
}

# 3. Create DataFrame and Export
df = pd.DataFrame(fsu_technical_data)
output_path = os.path.join(output_dir, "simulated_fsu_technical_data.csv")
df.to_csv(output_path, index=False)

print(f"Successfully generated {output_path} with {len(df)} records.")
print(f"Columns included: {df.columns.tolist()}")
```
